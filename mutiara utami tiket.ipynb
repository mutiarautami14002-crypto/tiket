```python

kode = input("Masukkan kode tiket (REG/PREM/VIP): ").upper()

match kode:
    case "REG":
        nama = "Reguler"
        harga = 10000
    case "PREM":
        nama = "Premiere"
        harga = 8000
    case "VIP":
        nama = "VIP"
        harga = 3000
    case _:
        nama = "Tidak ditemukan"
        harga = 0

print("Nama Tiket :", nama)
print("Harga      :", harga)
```
